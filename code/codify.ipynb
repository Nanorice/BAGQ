## 1 · Stars and bars, checked by brute force

`n` identical balls into `k` labelled bins.
Closed form `math.comb(n+k-1, k-1)`; brute force enumerates `itertools.product`.
Assert they agree for every `n <= 8`, `k <= 4`.

Docstring must state both complexities **and why brute force is fine as a test
and unacceptable as an implementation.**


```python
def factorial(n: int) -> int:
    """Compute n! (factorial of n)."""
    if n == 0 or n == 1:
        return 1
    result = 1
    for i in range(2, n + 1):
        result *= i
    return result

def factorial_recursive(n: int) -> int:
    """Compute n! (factorial of n) recursively."""
    if n == 0 or n == 1:
        return 1
    return n * factorial_recursive(n - 1)

assert factorial(5) == 120
assert factorial_recursive(5) == 120
```

## 1 · Geometric expectation

Mean number of `Bernoulli(p)` attempts until the first success. Verify simulated mean ~ `1/p`.


```python
total_trial = 0
for _ in range(1000):
    trial = 0
    success = 0
    while success < 1:
        trial += 1
        if random.random() < 0.2:
            success += 1
    total_trial += trial

simulated_mean = total_trial / 1000
theoretical_mean = 1/0.2

print(f"Simulated mean: {simulated_mean}")
print(f"Theoretical mean: {theoretical_mean}")
```

## 2 · Expected flips until two heads in a row

Mean fair-coin flips until `HH`. Verify the mean ~ `6`. (Interview favourite — mandatory.)


```python
total_flip = 0
for _ in range(1000):
    flips = 0
    HH = 0
    while HH < 2:
        flips += 1
        if random.random() < 0.5:
            HH += 1
        else:
            HH = 0
    total_flip += flips

simulated_mean = total_flip / 1000
theoretical_mean = 6

print(f"Simulated mean: {simulated_mean}")
print(f"Theoretical mean: {theoretical_mean}")
```
